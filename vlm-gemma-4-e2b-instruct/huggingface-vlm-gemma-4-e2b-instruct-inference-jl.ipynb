# Gemma 4 E2B Instruct - Inference Demo
This notebook demonstrates inference with the Gemma 4 E2B model.

```python
from sagemaker.predictor import retrieve_default
import json
```

```python
endpoint_name = "!!!name!!!"
predictor = retrieve_default(endpoint_name)
```

```python
payload = {
    "messages": [{"role": "user", "content": "What is deep learning? Answer in one sentence."}],
    "max_tokens": 128
}
response = predictor.predict(payload, initial_args={"ContentType": "application/json", "Accept": "application/json"})
print(json.dumps(response, indent=2) if isinstance(response, dict) else response)
```

# Image Generation: How to run inference on the endpoint you have created?

```python
import base64
import json
import boto3
from IPython.display import Image, display
```

### Generate an image from text

```python
endpoint_name = '!!!name!!!'

sm_runtime = boto3.client('sagemaker-runtime')

payload = {
    "model": "/opt/ml/model",
    "prompt": "A sunset over mountains with dramatic clouds",
    "n": 1,
    "size": "512x512"
}

response = sm_runtime.invoke_endpoint(
    EndpointName=endpoint_name,
    ContentType='application/json',
    CustomAttributes='route=/v1/images/generations',
    Body=json.dumps(payload),
)

result = json.loads(response['Body'].read())
print(f"Generated {len(result['data'])} image(s)")
```

### Display the generated image

```python
img_b64 = result['data'][0]['b64_json']
img_bytes = base64.b64decode(img_b64)
display(Image(data=img_bytes))
print(f'Image size: {len(img_bytes)} bytes')
```

### Save to file

```python
with open('generated_image.png', 'wb') as f:
    f.write(img_bytes)
print('Saved to generated_image.png')
```
